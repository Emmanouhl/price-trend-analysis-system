**MARKET PRICE FORECASTING SYSTEM**
=

This project builds a price forecasting system for Maize (Kano Central Market) and Rice (Makurdi International Market) using historical data and time-series to help traders and farmers make better market decisions.

*A Case Study of the Maize Commodity in Kano Central Market*
=

1. *Variable for information provided*

```python
commdity = "Maize"
market_location = "Kano Central Market"
yesterday_price = 58000
today_price = 61500
expected_price = 60000
quantity = 420
additional_supply = 80
minimum_supply = 450
market_open = True
```

2. *To Calculate Price Difference* 

```python
#To get the difference in commoity price
price_difference = today_price - yesterday_price
print(f"Price Difference : ₦{price_difference}")
```

3. *To Calculate the Percentage Price Change*

```python
percentage_change = (price_difference / yesterday_price) * 100
print(f"Percentage Price Change : {percentage_change:.2f}%")
```

4. *To Update Available Supply*

```python
#Since initial quantity is 420bags
quantity += additional_supply
print(f"Updated Quantity Available : {quantity}")
```

5. *To Check If Price Reached*

```python
price_reached = today_price >= expected_price
print(f"Expect Market Price Reached : {price_reached}")
```

6. *To Check if Supply is Sufficient*

```python
#Since updated available quantity is 500bags
supply_is_sufficient = quantity >= minimum_supply
print(f"Supply Sufficient : {supply_is_sufficient}")
```

7. *To Check if Market Conditions are Favorable*

```python
#Market is favorable if: 
#The market is opened and expected market is reached
market_favorable = (market_open and price_reached)
print(f"Market is favorable : {market_favorable}")
```

8. *Determine if Market Monitoring is Required*

```python
#Additinal monitoring required if:
#Market condition not favorable or supply not sufficient
monitoring_needed = not market_favorable or not supply_is_sufficient
print(f"Additional monitoring needed : {monitoring_needed}")
```

**THIS SECTION OF THE PROGRAM IS TO DISPLAY THE REPORT**
